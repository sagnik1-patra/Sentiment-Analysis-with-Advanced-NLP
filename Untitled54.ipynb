```python
# ======================================================================
# SENTIMENT ANALYSIS WITH ADVANCED NLP
# COMPLETE FIXED VERSION
# ======================================================================

# Install once if required:
# !pip install tensorflow numpy pandas matplotlib scikit-learn nltk gensim pyyaml

# ======================================================================
# 1. IMPORTS
# ======================================================================

import os
import re
import gc
import json
import yaml
import pickle
import random
import zipfile
import warnings
from pathlib import Path
from collections import Counter

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, regularizers

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score
)
from sklearn.utils.class_weight import compute_class_weight

import nltk
from nltk.corpus import stopwords

from gensim.models import Word2Vec


# ======================================================================
# 2. PROJECT DIRECTORY
# ======================================================================

PROJECT_DIR = Path(
    r"C:\Users\sagni\Downloads\Sentiment Analysis with Advanced NLP"
)

PLOTS_DIR = PROJECT_DIR / "plots"
MODELS_DIR = PROJECT_DIR / "models"
RESULTS_DIR = PROJECT_DIR / "results"

PROJECT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

PLOTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MODELS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

ZIP_PATH = (
    PROJECT_DIR.parent /
    "Sentiment_Analysis_with_Advanced_NLP.zip"
)


# ======================================================================
# 3. CONFIGURATION
# ======================================================================

SEED = 42

VOCAB_SIZE = 10000

MAX_LENGTH = 250

EMBEDDING_DIM = 64

BATCH_SIZE = 128

EPOCHS = 6

# Change to None to use all 50,000 reviews.
MAX_SAMPLES = 30000

TOP_ENSEMBLE_MODELS = 3


# ======================================================================
# 4. REPRODUCIBILITY
# ======================================================================

os.environ[
    "PYTHONHASHSEED"
] = str(SEED)

random.seed(
    SEED
)

np.random.seed(
    SEED
)

tf.random.set_seed(
    SEED
)


# ======================================================================
# 5. NLTK STOPWORDS
# ======================================================================

try:

    STOPWORDS = set(
        stopwords.words(
            "english"
        )
    )

except LookupError:

    nltk.download(
        "stopwords",
        quiet=True
    )

    STOPWORDS = set(
        stopwords.words(
            "english"
        )
    )


print(
    "=" * 70
)

print(
    "SENTIMENT ANALYSIS WITH ADVANCED NLP"
)

print(
    "=" * 70
)

print(
    "\nProject directory:"
)

print(
    PROJECT_DIR
)


# ======================================================================
# 6. LOAD IMDB DATASET
# ======================================================================

print(
    "\nLoading IMDB dataset..."
)

(
    (
        x_train_original,
        y_train_original
    ),
    (
        x_test_original,
        y_test_original
    )
) = tf.keras.datasets.imdb.load_data(
    num_words=VOCAB_SIZE
)


all_sequences = (
    list(
        x_train_original
    )
    +
    list(
        x_test_original
    )
)


all_labels = np.concatenate(
    [
        y_train_original,
        y_test_original
    ]
)


print(
    "Available reviews:",
    len(
        all_sequences
    )
)


# ======================================================================
# 7. RECONSTRUCT TEXT
# ======================================================================

word_index = (
    tf.keras.datasets.imdb
    .get_word_index()
)


reverse_word_index = {
    value + 3: key

    for key, value
    in word_index.items()
}


reverse_word_index[0] = (
    "<PAD>"
)

reverse_word_index[1] = (
    "<START>"
)

reverse_word_index[2] = (
    "<UNK>"
)

reverse_word_index[3] = (
    "<UNUSED>"
)


def decode_review(
    sequence
):

    words = []

    for index in sequence:

        word = (
            reverse_word_index
            .get(
                index,
                "<UNK>"
            )
        )

        words.append(
            word
        )

    return " ".join(
        words
    )


print(
    "\nDecoding reviews..."
)


all_texts = np.asarray(
    [
        decode_review(
            sequence
        )

        for sequence
        in all_sequences
    ],

    dtype=object
)


# ======================================================================
# 8. BALANCED SUBSAMPLING
# ======================================================================

if MAX_SAMPLES is not None:

    rng = (
        np.random
        .default_rng(
            SEED
        )
    )


    negative_indices = np.where(
        all_labels == 0
    )[0]


    positive_indices = np.where(
        all_labels == 1
    )[0]


    samples_per_class = (
        MAX_SAMPLES // 2
    )


    negative_selected = (
        rng.choice(
            negative_indices,

            size=min(
                samples_per_class,
                len(
                    negative_indices
                )
            ),

            replace=False
        )
    )


    positive_selected = (
        rng.choice(
            positive_indices,

            size=min(
                samples_per_class,
                len(
                    positive_indices
                )
            ),

            replace=False
        )
    )


    selected_indices = (
        np.concatenate(
            [
                negative_selected,
                positive_selected
            ]
        )
    )


    rng.shuffle(
        selected_indices
    )


    all_texts = (
        all_texts[
            selected_indices
        ]
    )


    all_labels = (
        all_labels[
            selected_indices
        ]
    )


all_labels = np.asarray(
    all_labels,
    dtype=np.int32
)


print(
    "Reviews used:",
    len(
        all_texts
    )
)


# ======================================================================
# 9. TEXT CLEANING
# ======================================================================

def clean_text(
    text,
    remove_stopwords=False
):

    text = str(
        text
    ).lower()


    # Remove HTML
    text = re.sub(
        r"<[^>]+>",
        " ",
        text
    )


    # Remove URLs
    text = re.sub(
        r"http\S+|www\.\S+",
        " ",
        text
    )


    # Remove special characters
    text = re.sub(
        r"[^a-z\s']",
        " ",
        text
    )


    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()


    words = (
        text.split()
    )


    special_tokens = {
        "pad",
        "start",
        "unk",
        "unused"
    }


    words = [
        word

        for word
        in words

        if word
        not in special_tokens
    ]


    if remove_stopwords:

        words = [
            word

            for word
            in words

            if word
            not in STOPWORDS
        ]


    return " ".join(
        words
    )


print(
    "\nCleaning reviews..."
)


cleaned_texts = np.asarray(
    [
        clean_text(
            text
        )

        for text
        in all_texts
    ],

    dtype=object
)


cleaned_without_stopwords = (
    np.asarray(
        [
            clean_text(
                text,
                remove_stopwords=True
            )

            for text
            in all_texts
        ],

        dtype=object
    )
)


# ======================================================================
# 10. SAVE PREPROCESSING SAMPLES
# ======================================================================

preprocessing_df = (
    pd.DataFrame(
        {
            "original_text":
                all_texts[:100],

            "cleaned_text":
                cleaned_texts[:100],

            "without_stopwords":
                cleaned_without_stopwords[
                    :100
                ],

            "sentiment":
                all_labels[:100]
        }
    )
)


preprocessing_df.to_csv(
    RESULTS_DIR /
    "preprocessing_samples.csv",

    index=False
)


# ======================================================================
# 11. CLASS DISTRIBUTION
# ======================================================================

class_counts = (
    pd.Series(
        all_labels
    )
    .value_counts()
    .sort_index()
)


plt.figure(
    figsize=(
        7,
        5
    )
)


plt.bar(
    [
        "Negative",
        "Positive"
    ],

    class_counts.values
)


plt.title(
    "Sentiment Class Distribution"
)


plt.xlabel(
    "Sentiment"
)


plt.ylabel(
    "Number of Reviews"
)


plt.tight_layout()


plt.savefig(
    PLOTS_DIR /
    "01_class_distribution.png",

    dpi=120
)


plt.show()

plt.close()


# ======================================================================
# 12. REVIEW LENGTH DISTRIBUTION
# ======================================================================

review_lengths = np.asarray(
    [
        len(
            text.split()
        )

        for text
        in cleaned_texts
    ]
)


plt.figure(
    figsize=(
        9,
        5
    )
)


plt.hist(
    review_lengths,
    bins=50
)


plt.axvline(
    MAX_LENGTH,

    linestyle="--",

    label=(
        f"Maximum sequence = "
        f"{MAX_LENGTH}"
    )
)


plt.title(
    "Review Length Distribution"
)


plt.xlabel(
    "Words per Review"
)


plt.ylabel(
    "Frequency"
)


plt.legend()


plt.tight_layout()


plt.savefig(
    PLOTS_DIR /
    "02_review_length_distribution.png",

    dpi=120
)


plt.show()

plt.close()


# ======================================================================
# 13. DATASET STATISTICS
# ======================================================================

dataset_statistics = {

    "total_samples":
        int(
            len(
                cleaned_texts
            )
        ),

    "negative_samples":
        int(
            np.sum(
                all_labels == 0
            )
        ),

    "positive_samples":
        int(
            np.sum(
                all_labels == 1
            )
        ),

    "average_review_length":
        float(
            np.mean(
                review_lengths
            )
        ),

    "median_review_length":
        float(
            np.median(
                review_lengths
            )
        ),

    "maximum_review_length":
        int(
            np.max(
                review_lengths
            )
        ),

    "sequence_length":
        MAX_LENGTH,

    "maximum_vocabulary":
        VOCAB_SIZE
}


with open(
    RESULTS_DIR /
    "dataset_statistics.json",

    "w",

    encoding="utf-8"
) as file:

    json.dump(
        dataset_statistics,
        file,
        indent=4
    )


# ======================================================================
# 14. WORD FREQUENCY
# ======================================================================

all_words = []


for text in cleaned_texts:

    all_words.extend(
        text.split()
    )


word_counter = Counter(
    all_words
)


top_words = (
    word_counter
    .most_common(
        20
    )
)


top_word_names = [
    word

    for word, count
    in top_words
]


top_word_counts = [
    count

    for word, count
    in top_words
]


plt.figure(
    figsize=(
        11,
        6
    )
)


plt.bar(
    top_word_names,
    top_word_counts
)


plt.xticks(
    rotation=60,
    ha="right"
)


plt.title(
    "Top 20 Most Frequent Words"
)


plt.xlabel(
    "Word"
)


plt.ylabel(
    "Frequency"
)


plt.tight_layout()


plt.savefig(
    PLOTS_DIR /
    "03_top_words.png",

    dpi=120
)


plt.show()

plt.close()


# ======================================================================
# 15. STOPWORD COMPARISON
#
# FIX:
# We DO NOT use tick_labels= here.
# We also avoid labels= for maximum Matplotlib compatibility.
# ======================================================================

with_stopwords_lengths = (
    np.asarray(
        [
            len(
                text.split()
            )

            for text
            in cleaned_texts
        ]
    )
)


without_stopwords_lengths = (
    np.asarray(
        [
            len(
                text.split()
            )

            for text
            in cleaned_without_stopwords
        ]
    )
)


plt.figure(
    figsize=(
        8,
        5
    )
)


plt.boxplot(
    [
        with_stopwords_lengths,
        without_stopwords_lengths
    ],

    showfliers=False
)


# Version-safe labels
plt.xticks(
    [
        1,
        2
    ],

    [
        "With Stopwords",
        "Without Stopwords"
    ]
)


plt.ylabel(
    "Review Length"
)


plt.title(
    "Effect of Stopword Removal"
)


plt.tight_layout()


plt.savefig(
    PLOTS_DIR /
    "04_stopword_comparison.png",

    dpi=120
)


plt.show()

plt.close()


# ======================================================================
# 16. TRAIN / VALIDATION / TEST SPLIT
# 70 / 15 / 15
# ======================================================================

(
    X_train_text,
    X_temp_text,
    y_train,
    y_temp
) = train_test_split(

    cleaned_texts,
    all_labels,

    test_size=0.30,

    random_state=SEED,

    stratify=all_labels
)


(
    X_val_text,
    X_test_text,
    y_val,
    y_test
) = train_test_split(

    X_temp_text,
    y_temp,

    test_size=0.50,

    random_state=SEED,

    stratify=y_temp
)


print(
    "\nDataset split:"
)


print(
    "Training:",
    len(
        X_train_text
    )
)


print(
    "Validation:",
    len(
        X_val_text
    )
)


print(
    "Testing:",
    len(
        X_test_text
    )
)


# ======================================================================
# 17. TEXT VECTORIZATION
# ======================================================================

vectorizer = (
    layers.TextVectorization(

        max_tokens=
            VOCAB_SIZE,

        output_mode=
            "int",

        output_sequence_length=
            MAX_LENGTH,

        standardize=
            None
    )
)


print(
    "\nAdapting text vectorizer..."
)


vectorizer.adapt(
    X_train_text
)


X_train = (
    vectorizer(
        X_train_text
    )
    .numpy()
)


X_val = (
    vectorizer(
        X_val_text
    )
    .numpy()
)


X_test = (
    vectorizer(
        X_test_text
    )
    .numpy()
)


vocabulary = (
    vectorizer
    .get_vocabulary()
)


actual_vocab_size = (
    len(
        vocabulary
    )
)


print(
    "Actual vocabulary size:",
    actual_vocab_size
)


# ======================================================================
# 18. SAVE VOCABULARY
# ======================================================================

with open(
    RESULTS_DIR /
    "vocabulary.json",

    "w",

    encoding="utf-8"
) as file:

    json.dump(
        vocabulary,
        file
    )


# ======================================================================
# 19. CHARACTER-LEVEL TOKENIZATION
# ======================================================================

character_vectorizer = (
    layers.TextVectorization(

        max_tokens=100,

        output_mode="int",

        output_sequence_length=500,

        split="character",

        standardize=None
    )
)


character_vectorizer.adapt(
    X_train_text[
        :5000
    ]
)


character_example = (
    character_vectorizer(
        [
            X_train_text[0]
        ]
    )
    .numpy()[0]
)


character_information = {

    "sample":
        str(
            X_train_text[0][
                :500
            ]
        ),

    "first_100_character_tokens":
        character_example[
            :100
        ].tolist(),

    "character_vocabulary_size":
        len(
            character_vectorizer
            .get_vocabulary()
        )
}


with open(
    RESULTS_DIR /
    "character_tokenization.json",

    "w",

    encoding="utf-8"
) as file:

    json.dump(
        character_information,
        file,
        indent=4
    )


# ======================================================================
# 20. WORD2VEC
# ======================================================================

print(
    "\nTraining custom Word2Vec..."
)


tokenized_reviews = [
    text.split()

    for text
    in X_train_text
]


word2vec_model = Word2Vec(

    sentences=
        tokenized_reviews,

    vector_size=
        EMBEDDING_DIM,

    window=5,

    min_count=2,

    workers=4,

    sg=1,

    epochs=5,

    seed=SEED
)


word2vec_matrix = (
    np.random.normal(

        loc=0.0,

        scale=0.05,

        size=(
            actual_vocab_size,
            EMBEDDING_DIM
        )

    )
    .astype(
        np.float32
    )
)


word2vec_matrix[0] = 0


word2vec_hits = 0


for index, word in enumerate(
    vocabulary
):

    if word in word2vec_model.wv:

        word2vec_matrix[
            index
        ] = (
            word2vec_model
            .wv[word]
        )

        word2vec_hits += 1


print(
    "Word2Vec vocabulary matches:",
    word2vec_hits
)


# ======================================================================
# 21. OPTIONAL GLOVE EMBEDDINGS
# ======================================================================

GLOVE_PATH = (
    PROJECT_DIR /
    "glove.6B.50d.txt"
)


GLOVE_DIM = 50


def build_glove_matrix():

    matrix = (
        np.random.normal(

            loc=0.0,

            scale=0.05,

            size=(
                actual_vocab_size,
                GLOVE_DIM
            )
        )
        .astype(
            np.float32
        )
    )


    matrix[0] = 0


    hits = 0


    if not GLOVE_PATH.exists():

        print(
            "\nGloVe file not found."
        )

        print(
            "The project will continue "
            "without external GloVe."
        )

        print(
            "Optional location:"
        )

        print(
            GLOVE_PATH
        )

        return (
            matrix,
            hits
        )


    print(
        "\nLoading GloVe embeddings..."
    )


    vocabulary_lookup = {

        word: index

        for index, word
        in enumerate(
            vocabulary
        )
    }


    with open(
        GLOVE_PATH,

        "r",

        encoding="utf-8"
    ) as file:

        for line in file:

            values = (
                line
                .rstrip()
                .split()
            )


            if len(values) != (
                GLOVE_DIM + 1
            ):

                continue


            word = values[0]


            if word not in (
                vocabulary_lookup
            ):

                continue


            vector = np.asarray(
                values[1:],
                dtype=np.float32
            )


            index = (
                vocabulary_lookup[
                    word
                ]
            )


            matrix[
                index
            ] = vector


            hits += 1


    return (
        matrix,
        hits
    )


(
    glove_matrix,
    glove_hits
) = build_glove_matrix()


embedding_information = {

    "word2vec_dimension":
        EMBEDDING_DIM,

    "word2vec_matches":
        int(
            word2vec_hits
        ),

    "glove_dimension":
        GLOVE_DIM,

    "glove_matches":
        int(
            glove_hits
        ),

    "glove_loaded":
        bool(
            GLOVE_PATH.exists()
        )
}


with open(
    RESULTS_DIR /
    "embedding_comparison.json",

    "w",

    encoding="utf-8"
) as file:

    json.dump(
        embedding_information,
        file,
        indent=4
    )


# ======================================================================
# 22. EMBEDDING COVERAGE GRAPH
# ======================================================================

plt.figure(
    figsize=(
        7,
        5
    )
)


plt.bar(
    [
        "Word2Vec",
        "GloVe"
    ],

    [
        word2vec_hits,
        glove_hits
    ]
)


plt.title(
    "Embedding Vocabulary Coverage"
)


plt.ylabel(
    "Matched Vocabulary Words"
)


plt.tight_layout()


plt.savefig(
    PLOTS_DIR /
    "05_embedding_coverage.png",

    dpi=120
)


plt.show()

plt.close()


# ======================================================================
# 23. CLASS WEIGHTS
# ======================================================================

unique_classes = np.unique(
    y_train
)


weights = (
    compute_class_weight(

        class_weight=
            "balanced",

        classes=
            unique_classes,

        y=
            y_train
    )
)


class_weights = {

    int(
        class_id
    ):
        float(
            weight
        )

    for class_id, weight
    in zip(
        unique_classes,
        weights
    )
}


print(
    "\nClass weights:"
)

print(
    class_weights
)


# ======================================================================
# 24. CUSTOM ATTENTION LAYER
# ======================================================================

@tf.keras.utils.register_keras_serializable(
    package="SentimentNLP"
)
class AttentionLayer(
    layers.Layer
):

    def __init__(
        self,
        **kwargs
    ):

        super().__init__(
            **kwargs
        )


    def build(
        self,
        input_shape
    ):

        self.W = (
            self.add_weight(

                name=
                    "attention_weight",

                shape=(
                    input_shape[-1],
                    1
                ),

                initializer=
                    "glorot_uniform",

                trainable=True
            )
        )


        self.b = (
            self.add_weight(

                name=
                    "attention_bias",

                shape=(1,),

                initializer=
                    "zeros",

                trainable=True
            )
        )


        super().build(
            input_shape
        )


    def call(
        self,
        inputs
    ):

        scores = tf.squeeze(

            tf.tanh(

                tf.matmul(
                    inputs,
                    self.W
                )
                +
                self.b
            ),

            axis=-1
        )


        attention_weights = (
            tf.nn.softmax(
                scores,
                axis=1
            )
        )


        context = (
            tf.reduce_sum(

                inputs
                *
                tf.expand_dims(
                    attention_weights,
                    axis=-1
                ),

                axis=1
            )
        )


        return context


# ======================================================================
# 25. EMBEDDING LAYER
# ======================================================================

def create_embedding():

    return layers.Embedding(

        input_dim=
            actual_vocab_size,

        output_dim=
            EMBEDDING_DIM,

        weights=[
            word2vec_matrix
        ],

        trainable=True,

        name=
            "word2vec_embedding"
    )


# ======================================================================
# 26. BIDIRECTIONAL LSTM
# ======================================================================

def build_bilstm():

    inputs = layers.Input(
        shape=(
            MAX_LENGTH,
        )
    )


    x = create_embedding()(
        inputs
    )


    x = layers.Bidirectional(

        layers.LSTM(
            64,
            dropout=0.20
        )

    )(x)


    x = layers.Dropout(
        0.50
    )(x)


    x = layers.Dense(

        64,

        activation=
            "relu",

        kernel_regularizer=
            regularizers.l2(
                1e-4
            )

    )(x)


    x = layers.Dropout(
        0.30
    )(x)


    outputs = layers.Dense(

        1,

        activation=
            "sigmoid"

    )(x)


    return models.Model(

        inputs=
            inputs,

        outputs=
            outputs,

        name=
            "bilstm"
    )


# ======================================================================
# 27. GRU + ATTENTION
# ======================================================================

def build_gru_attention():

    inputs = layers.Input(
        shape=(
            MAX_LENGTH,
        )
    )


    x = create_embedding()(
        inputs
    )


    x = layers.GRU(

        64,

        return_sequences=
            True,

        dropout=
            0.20

    )(x)


    x = AttentionLayer(
        name=
            "attention"
    )(x)


    x = layers.Dense(

        64,

        activation=
            "relu"

    )(x)


    x = layers.Dropout(
        0.40
    )(x)


    outputs = layers.Dense(

        1,

        activation=
            "sigmoid"

    )(x)


    return models.Model(

        inputs=
            inputs,

        outputs=
            outputs,

        name=
            "gru_attention"
    )


# ======================================================================
# 28. TEXT CNN
# ======================================================================

def build_text_cnn():

    inputs = layers.Input(
        shape=(
            MAX_LENGTH,
        )
    )


    x = create_embedding()(
        inputs
    )


    x = layers.Conv1D(

        128,

        kernel_size=5,

        padding=
            "same",

        activation=
            "relu"

    )(x)


    x = layers.MaxPooling1D(
        pool_size=2
    )(x)


    x = layers.Conv1D(

        64,

        kernel_size=3,

        padding=
            "same",

        activation=
            "relu"

    )(x)


    x = (
        layers.GlobalMaxPooling1D()(
            x
        )
    )


    x = layers.Dense(

        64,

        activation=
            "relu"

    )(x)


    x = layers.Dropout(
        0.40
    )(x)


    outputs = layers.Dense(

        1,

        activation=
            "sigmoid"

    )(x)


    return models.Model(

        inputs=
            inputs,

        outputs=
            outputs,

        name=
            "text_cnn"
    )


# ======================================================================
# 29. CNN-LSTM
# ======================================================================

def build_cnn_lstm():

    inputs = layers.Input(
        shape=(
            MAX_LENGTH,
        )
    )


    x = create_embedding()(
        inputs
    )


    x = layers.Conv1D(

        96,

        kernel_size=5,

        padding=
            "same",

        activation=
            "relu"

    )(x)


    x = layers.MaxPooling1D(
        pool_size=2
    )(x)


    x = layers.LSTM(

        48,

        dropout=
            0.20

    )(x)


    x = layers.Dense(

        64,

        activation=
            "relu"

    )(x)


    x = layers.Dropout(
        0.40
    )(x)


    outputs = layers.Dense(

        1,

        activation=
            "sigmoid"

    )(x)


    return models.Model(

        inputs=
            inputs,

        outputs=
            outputs,

        name=
            "cnn_lstm"
    )


# ======================================================================
# 30. VALIDATION F1 CALLBACK
# ======================================================================

class ValidationF1Callback(
    callbacks.Callback
):

    def __init__(
        self,
        X_validation,
        y_validation
    ):

        super().__init__()

        self.X_validation = (
            X_validation
        )

        self.y_validation = (
            y_validation
        )

        self.best_f1 = -1.0

        self.best_weights = None


    def on_epoch_end(
        self,
        epoch,
        logs=None
    ):

        probabilities = (
            self.model.predict(

                self.X_validation,

                batch_size=
                    BATCH_SIZE,

                verbose=0

            ).ravel()
        )


        predictions = (
            probabilities
            >= 0.5
        ).astype(
            np.int32
        )


        current_f1 = f1_score(

            self.y_validation,

            predictions,

            zero_division=0
        )


        print(
            f"val_f1_custom: "
            f"{current_f1:.4f}"
        )


        if current_f1 > (
            self.best_f1
        ):

            self.best_f1 = (
                current_f1
            )

            self.best_weights = (
                self.model
                .get_weights()
            )


    def on_train_end(
        self,
        logs=None
    ):

        if self.best_weights is not None:

            self.model.set_weights(
                self.best_weights
            )


# ======================================================================
# 31. TRAIN MODEL FUNCTION
# ======================================================================

def train_model(
    model,
    model_name
):

    print(
        "\n" +
        "=" * 70
    )

    print(
        "TRAINING:",
        model_name
    )

    print(
        "=" * 70
    )


    model.compile(

        optimizer=
            tf.keras.optimizers.Adam(
                learning_rate=
                    0.001
            ),

        loss=
            "binary_crossentropy",

        metrics=[
            "accuracy",

            tf.keras.metrics.Precision(
                name="precision"
            ),

            tf.keras.metrics.Recall(
                name="recall"
            )
        ]
    )


    model.summary()


    validation_f1 = (
        ValidationF1Callback(
            X_val,
            y_val
        )
    )


    early_stopping = (
        callbacks.EarlyStopping(

            monitor=
                "val_loss",

            patience=2,

            restore_best_weights=
                False
        )
    )


    reduce_lr = (
        callbacks.ReduceLROnPlateau(

            monitor=
                "val_loss",

            factor=0.5,

            patience=1,

            min_lr=1e-5,

            verbose=1
        )
    )


    history = model.fit(

        X_train,
        y_train,

        validation_data=(
            X_val,
            y_val
        ),

        epochs=
            EPOCHS,

        batch_size=
            BATCH_SIZE,

        class_weight=
            class_weights,

        callbacks=[
            validation_f1,
            early_stopping,
            reduce_lr
        ],

        verbose=1
    )


    if (
        validation_f1
        .best_weights
        is not None
    ):

        model.set_weights(
            validation_f1
            .best_weights
        )


    weight_path = (
        MODELS_DIR /
        (
            f"{model_name}"
            ".weights.h5"
        )
    )


    model.save_weights(
        weight_path
    )


    history_dictionary = {

        key: [
            float(value)

            for value
            in values
        ]

        for key, values
        in history.history.items()
    }


    with open(

        RESULTS_DIR /
        (
            f"{model_name}"
            "_history.json"
        ),

        "w",

        encoding="utf-8"

    ) as file:

        json.dump(
            history_dictionary,
            file,
            indent=4
        )


    return (
        model,
        history_dictionary,
        validation_f1.best_f1
    )


# ======================================================================
# 32. MODEL BUILDERS
# ======================================================================

model_builders = {

    "bilstm":
        build_bilstm,

    "gru_attention":
        build_gru_attention,

    "text_cnn":
        build_text_cnn,

    "cnn_lstm":
        build_cnn_lstm
}


trained_models = {}

histories = {}

validation_f1_scores = {}


# ======================================================================
# 33. TRAIN MODELS
# ======================================================================

for model_name, builder in (
    model_builders.items()
):

    gc.collect()


    current_model = (
        builder()
    )


    (
        current_model,
        current_history,
        current_best_f1
    ) = train_model(

        current_model,

        model_name
    )


    trained_models[
        model_name
    ] = current_model


    histories[
        model_name
    ] = current_history


    validation_f1_scores[
        model_name
    ] = float(
        current_best_f1
    )


# ======================================================================
# 34. TRAINING CURVES
# ======================================================================

plot_number = 6


for model_name, history in (
    histories.items()
):

    # Accuracy
    plt.figure(
        figsize=(
            8,
            5
        )
    )


    plt.plot(
        history[
            "accuracy"
        ],
        label=
            "Training Accuracy"
    )


    plt.plot(
        history[
            "val_accuracy"
        ],
        label=
            "Validation Accuracy"
    )


    plt.title(
        f"{model_name} Accuracy"
    )


    plt.xlabel(
        "Epoch"
    )


    plt.ylabel(
        "Accuracy"
    )


    plt.legend()


    plt.tight_layout()


    plt.savefig(

        PLOTS_DIR /
        (
            f"{plot_number:02d}_"
            f"{model_name}_accuracy.png"
        ),

        dpi=120
    )


    plt.show()

    plt.close()


    plot_number += 1


    # Loss
    plt.figure(
        figsize=(
            8,
            5
        )
    )


    plt.plot(
        history[
            "loss"
        ],
        label=
            "Training Loss"
    )


    plt.plot(
        history[
            "val_loss"
        ],
        label=
            "Validation Loss"
    )


    plt.title(
        f"{model_name} Loss"
    )


    plt.xlabel(
        "Epoch"
    )


    plt.ylabel(
        "Loss"
    )


    plt.legend()


    plt.tight_layout()


    plt.savefig(

        PLOTS_DIR /
        (
            f"{plot_number:02d}_"
            f"{model_name}_loss.png"
        ),

        dpi=120
    )


    plt.show()

    plt.close()


    plot_number += 1


# ======================================================================
# 35. EVALUATE MODELS
# ======================================================================

model_results = []

test_probabilities = {}

test_predictions = {}


for model_name, model in (
    trained_models.items()
):

    print(
        "\nEvaluating:",
        model_name
    )


    probabilities = (
        model.predict(

            X_test,

            batch_size=
                BATCH_SIZE,

            verbose=0

        ).ravel()
    )


    predictions = (
        probabilities
        >= 0.5
    ).astype(
        np.int32
    )


    test_probabilities[
        model_name
    ] = probabilities


    test_predictions[
        model_name
    ] = predictions


    accuracy = accuracy_score(
        y_test,
        predictions
    )


    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )


    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )


    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )


    roc_auc = roc_auc_score(
        y_test,
        probabilities
    )


    model_results.append(
        {
            "Model":
                model_name,

            "Accuracy":
                float(
                    accuracy
                ),

            "Precision":
                float(
                    precision
                ),

            "Recall":
                float(
                    recall
                ),

            "F1":
                float(
                    f1
                ),

            "ROC_AUC":
                float(
                    roc_auc
                )
        }
    )


# ======================================================================
# 36. CONFUSION MATRICES
# ======================================================================

for model_name in (
    trained_models.keys()
):

    predictions = (
        test_predictions[
            model_name
        ]
    )


    cm = confusion_matrix(
        y_test,
        predictions
    )


    plt.figure(
        figsize=(
            6,
            5
        )
    )


    plt.imshow(
        cm,
        interpolation=
            "nearest"
    )


    plt.colorbar()


    plt.title(
        f"{model_name} "
        "Confusion Matrix"
    )


    plt.xticks(
        [
            0,
            1
        ],
        [
            "Negative",
            "Positive"
        ]
    )


    plt.yticks(
        [
            0,
            1
        ],
        [
            "Negative",
            "Positive"
        ]
    )


    for row in range(2):

        for column in range(2):

            plt.text(

                column,
                row,

                str(
                    cm[
                        row,
                        column
                    ]
                ),

                horizontalalignment=
                    "center"
            )


    plt.xlabel(
        "Predicted"
    )


    plt.ylabel(
        "Actual"
    )


    plt.tight_layout()


    plt.savefig(

        PLOTS_DIR /
        (
            f"{plot_number:02d}_"
            f"{model_name}_"
            "confusion_matrix.png"
        ),

        dpi=120
    )


    plt.show()

    plt.close()


    plot_number += 1


# ======================================================================
# 37. ROC CURVES
# ======================================================================

for model_name in (
    trained_models.keys()
):

    probabilities = (
        test_probabilities[
            model_name
        ]
    )


    fpr, tpr, _ = (
        roc_curve(
            y_test,
            probabilities
        )
    )


    auc_score = (
        roc_auc_score(
            y_test,
            probabilities
        )
    )


    plt.figure(
        figsize=(
            7,
            5
        )
    )


    plt.plot(
        fpr,
        tpr,

        label=(
            f"AUC = "
            f"{auc_score:.4f}"
        )
    )


    plt.plot(
        [
            0,
            1
        ],

        [
            0,
            1
        ],

        linestyle="--"
    )


    plt.title(
        f"{model_name} ROC Curve"
    )


    plt.xlabel(
        "False Positive Rate"
    )


    plt.ylabel(
        "True Positive Rate"
    )


    plt.legend()


    plt.tight_layout()


    plt.savefig(

        PLOTS_DIR /
        (
            f"{plot_number:02d}_"
            f"{model_name}_roc_curve.png"
        ),

        dpi=120
    )


    plt.show()

    plt.close()


    plot_number += 1


# ======================================================================
# 38. MODEL RESULTS
# ======================================================================

results_dataframe = (
    pd.DataFrame(
        model_results
    )
)


results_dataframe = (
    results_dataframe
    .sort_values(
        "F1",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


print(
    "\nMODEL COMPARISON"
)


print(
    results_dataframe
)


results_dataframe.to_csv(

    RESULTS_DIR /
    "model_comparison.csv",

    index=False
)


# ======================================================================
# 39. ACCURACY COMPARISON
# ======================================================================

plt.figure(
    figsize=(
        9,
        5
    )
)


plt.bar(
    results_dataframe[
        "Model"
    ],

    results_dataframe[
        "Accuracy"
    ]
)


plt.ylim(
    0,
    1
)


plt.title(
    "Model Accuracy Comparison"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Accuracy"
)


plt.xticks(
    rotation=20
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "model_accuracy_comparison.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 40. F1 COMPARISON
# ======================================================================

plt.figure(
    figsize=(
        9,
        5
    )
)


plt.bar(
    results_dataframe[
        "Model"
    ],

    results_dataframe[
        "F1"
    ]
)


plt.ylim(
    0,
    1
)


plt.title(
    "Model F1 Score Comparison"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "F1 Score"
)


plt.xticks(
    rotation=20
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "model_f1_comparison.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 41. COMPLETE METRIC COMPARISON
# ======================================================================

metric_names = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC"
]


x_positions = np.arange(
    len(
        results_dataframe
    )
)


bar_width = 0.15


plt.figure(
    figsize=(
        12,
        6
    )
)


for metric_index, metric in enumerate(
    metric_names
):

    plt.bar(

        x_positions
        +
        metric_index
        *
        bar_width,

        results_dataframe[
            metric
        ],

        width=
            bar_width,

        label=
            metric
    )


plt.xticks(

    x_positions
    +
    2
    *
    bar_width,

    results_dataframe[
        "Model"
    ],

    rotation=20
)


plt.ylim(
    0,
    1
)


plt.ylabel(
    "Score"
)


plt.title(
    "Complete Model Performance Comparison"
)


plt.legend()


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "complete_model_comparison.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 42. ENSEMBLE
# ======================================================================

top_models = (
    results_dataframe
    .head(
        TOP_ENSEMBLE_MODELS
    )[
        "Model"
    ]
    .tolist()
)


print(
    "\nModels selected "
    "for ensemble:"
)


print(
    top_models
)


ensemble_probabilities = (
    np.mean(
        [
            test_probabilities[
                model_name
            ]

            for model_name
            in top_models
        ],

        axis=0
    )
)


ensemble_predictions = (
    ensemble_probabilities
    >= 0.5
).astype(
    np.int32
)


ensemble_accuracy = (
    accuracy_score(
        y_test,
        ensemble_predictions
    )
)


ensemble_precision = (
    precision_score(
        y_test,
        ensemble_predictions,
        zero_division=0
    )
)


ensemble_recall = (
    recall_score(
        y_test,
        ensemble_predictions,
        zero_division=0
    )
)


ensemble_f1 = (
    f1_score(
        y_test,
        ensemble_predictions,
        zero_division=0
    )
)


ensemble_auc = (
    roc_auc_score(
        y_test,
        ensemble_probabilities
    )
)


ensemble_result = {

    "Model":
        "ensemble_top_3",

    "Accuracy":
        float(
            ensemble_accuracy
        ),

    "Precision":
        float(
            ensemble_precision
        ),

    "Recall":
        float(
            ensemble_recall
        ),

    "F1":
        float(
            ensemble_f1
        ),

    "ROC_AUC":
        float(
            ensemble_auc
        )
}


print(
    "\nENSEMBLE RESULTS"
)


print(
    ensemble_result
)


# ======================================================================
# 43. FINAL COMPARISON
# ======================================================================

final_results_dataframe = (
    pd.concat(
        [
            results_dataframe,

            pd.DataFrame(
                [
                    ensemble_result
                ]
            )
        ],

        ignore_index=True
    )
)


final_results_dataframe.to_csv(

    RESULTS_DIR /
    "final_model_comparison.csv",

    index=False
)


plt.figure(
    figsize=(
        10,
        6
    )
)


plt.bar(
    final_results_dataframe[
        "Model"
    ],

    final_results_dataframe[
        "F1"
    ]
)


plt.ylim(
    0,
    1
)


plt.title(
    "Individual Models vs Ensemble"
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "F1 Score"
)


plt.xticks(
    rotation=25
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "ensemble_comparison.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 44. ENSEMBLE CONFUSION MATRIX
# ======================================================================

ensemble_cm = confusion_matrix(
    y_test,
    ensemble_predictions
)


plt.figure(
    figsize=(
        6,
        5
    )
)


plt.imshow(
    ensemble_cm,
    interpolation=
        "nearest"
)


plt.colorbar()


plt.title(
    "Ensemble Confusion Matrix"
)


plt.xticks(
    [
        0,
        1
    ],
    [
        "Negative",
        "Positive"
    ]
)


plt.yticks(
    [
        0,
        1
    ],
    [
        "Negative",
        "Positive"
    ]
)


for row in range(2):

    for column in range(2):

        plt.text(

            column,
            row,

            str(
                ensemble_cm[
                    row,
                    column
                ]
            ),

            horizontalalignment=
                "center"
        )


plt.xlabel(
    "Predicted"
)


plt.ylabel(
    "Actual"
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "ensemble_confusion_matrix.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 45. SAVE PREDICTIONS
# ======================================================================

prediction_dataframe = (
    pd.DataFrame(
        {
            "review":
                X_test_text,

            "actual_sentiment":
                y_test,

            "predicted_sentiment":
                ensemble_predictions,

            "positive_probability":
                ensemble_probabilities,

            "correct":
                (
                    ensemble_predictions
                    ==
                    y_test
                )
        }
    )
)


prediction_dataframe.to_csv(

    RESULTS_DIR /
    "predictions.csv.gz",

    index=False,

    compression=
        "gzip"
)


# ======================================================================
# 46. ERROR ANALYSIS
# ======================================================================

error_dataframe = (
    prediction_dataframe[
        prediction_dataframe[
            "correct"
        ] == False
    ]
    .copy()
)


error_dataframe[
    "confidence"
] = np.where(

    error_dataframe[
        "predicted_sentiment"
    ] == 1,

    error_dataframe[
        "positive_probability"
    ],

    1
    -
    error_dataframe[
        "positive_probability"
    ]
)


error_dataframe = (
    error_dataframe
    .sort_values(
        "confidence",
        ascending=False
    )
)


error_dataframe.head(
    200
).to_csv(

    RESULTS_DIR /
    "error_analysis.csv",

    index=False
)


# ======================================================================
# 47. ERROR CONFIDENCE
# ======================================================================

plt.figure(
    figsize=(
        8,
        5
    )
)


plt.hist(
    error_dataframe[
        "confidence"
    ],
    bins=30
)


plt.title(
    "Misclassification Confidence Distribution"
)


plt.xlabel(
    "Confidence"
)


plt.ylabel(
    "Misclassified Reviews"
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "error_confidence.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 48. CORRECT VS INCORRECT
# ======================================================================

correct_count = int(
    prediction_dataframe[
        "correct"
    ].sum()
)


incorrect_count = int(
    len(
        prediction_dataframe
    )
    -
    correct_count
)


plt.figure(
    figsize=(
        7,
        5
    )
)


plt.bar(
    [
        "Correct",
        "Incorrect"
    ],

    [
        correct_count,
        incorrect_count
    ]
)


plt.title(
    "Correct vs Incorrect Predictions"
)


plt.ylabel(
    "Number of Reviews"
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "correct_vs_incorrect.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 49. PREDICTION PROBABILITY
# ======================================================================

plt.figure(
    figsize=(
        8,
        5
    )
)


plt.hist(
    ensemble_probabilities,
    bins=40
)


plt.axvline(
    0.5,
    linestyle="--"
)


plt.title(
    "Prediction Probability Distribution"
)


plt.xlabel(
    "Positive Sentiment Probability"
)


plt.ylabel(
    "Frequency"
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "prediction_probability.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 50. REVIEW LENGTH VS ERROR
#
# SECOND BOXPLOT FIXED HERE TOO.
# ======================================================================

test_lengths = np.asarray(
    [
        len(
            text.split()
        )

        for text
        in X_test_text
    ]
)


correct_mask = (
    ensemble_predictions
    ==
    y_test
)


plt.figure(
    figsize=(
        8,
        5
    )
)


plt.boxplot(
    [
        test_lengths[
            correct_mask
        ],

        test_lengths[
            ~correct_mask
        ]
    ],

    showfliers=False
)


# Version-safe labels
plt.xticks(
    [
        1,
        2
    ],

    [
        "Correct",
        "Incorrect"
    ]
)


plt.title(
    "Review Length vs Prediction Correctness"
)


plt.ylabel(
    "Review Length"
)


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "review_length_error.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 51. WORD IMPORTANCE
# ======================================================================

def predict_single_text(
    model,
    text
):

    sequence = vectorizer(
        [
            text
        ]
    )


    probability = (
        model.predict(
            sequence,
            verbose=0
        )[0][0]
    )


    return float(
        probability
    )


def calculate_word_importance(
    model,
    text,
    maximum_words=25
):

    words = (
        text.split()[
            :maximum_words
        ]
    )


    if len(words) == 0:

        return (
            [],
            []
        )


    short_text = " ".join(
        words
    )


    original_probability = (
        predict_single_text(
            model,
            short_text
        )
    )


    importance_scores = []


    for index in range(
        len(
            words
        )
    ):

        modified_words = (
            words[:index]
            +
            words[
                index + 1:
            ]
        )


        modified_text = " ".join(
            modified_words
        )


        modified_probability = (
            predict_single_text(
                model,
                modified_text
            )
        )


        importance = abs(
            original_probability
            -
            modified_probability
        )


        importance_scores.append(
            importance
        )


    return (
        words,
        importance_scores
    )


interpretability_model = (
    trained_models[
        "gru_attention"
    ]
)


sample_text = str(
    X_test_text[0]
)


(
    importance_words,
    importance_scores
) = calculate_word_importance(

    interpretability_model,

    sample_text
)


if len(
    importance_words
) > 0:

    plt.figure(
        figsize=(
            12,
            6
        )
    )


    plt.bar(
        importance_words,
        importance_scores
    )


    plt.xticks(
        rotation=70,
        ha="right"
    )


    plt.title(
        "Word Importance Visualization"
    )


    plt.xlabel(
        "Word"
    )


    plt.ylabel(
        "Importance"
    )


    plt.tight_layout()


    plt.savefig(

        PLOTS_DIR /
        (
            f"{plot_number:02d}_"
            "word_importance.png"
        ),

        dpi=120
    )


    plt.show()

    plt.close()


    plot_number += 1


# ======================================================================
# 52. SENTIMENT PROBABILITY DISTRIBUTION
# ======================================================================

plt.figure(
    figsize=(
        8,
        5
    )
)


plt.hist(
    ensemble_probabilities[
        y_test == 0
    ],

    bins=30,

    alpha=0.6,

    label=
        "Actual Negative"
)


plt.hist(
    ensemble_probabilities[
        y_test == 1
    ],

    bins=30,

    alpha=0.6,

    label=
        "Actual Positive"
)


plt.title(
    "Probability Distribution by Actual Sentiment"
)


plt.xlabel(
    "Positive Probability"
)


plt.ylabel(
    "Frequency"
)


plt.legend()


plt.tight_layout()


plt.savefig(

    PLOTS_DIR /
    (
        f"{plot_number:02d}_"
        "sentiment_probability_distribution.png"
    ),

    dpi=120
)


plt.show()

plt.close()


plot_number += 1


# ======================================================================
# 53. CLASSIFICATION REPORT
# ======================================================================

classification_results = (
    classification_report(

        y_test,

        ensemble_predictions,

        target_names=[
            "Negative",
            "Positive"
        ],

        output_dict=True,

        zero_division=0
    )
)


with open(

    RESULTS_DIR /
    "classification_report.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(
        classification_results,
        file,
        indent=4
    )


# ======================================================================
# 54. CONFIG YAML
# ======================================================================

configuration = {

    "project":
        "Sentiment Analysis with Advanced NLP",

    "dataset":
        "IMDB Movie Reviews",

    "seed":
        SEED,

    "maximum_samples":
        MAX_SAMPLES,

    "vocabulary_size":
        VOCAB_SIZE,

    "sequence_length":
        MAX_LENGTH,

    "embedding_dimension":
        EMBEDDING_DIM,

    "batch_size":
        BATCH_SIZE,

    "epochs":
        EPOCHS,

    "training_split":
        0.70,

    "validation_split":
        0.15,

    "test_split":
        0.15,

    "models": [
        "Bidirectional LSTM",
        "GRU with Attention",
        "1D Text CNN",
        "CNN-LSTM"
    ],

    "optimizer":
        "Adam",

    "loss":
        "Binary Crossentropy",

    "ensemble_models":
        top_models
}


with open(

    PROJECT_DIR /
    "config.yaml",

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(
        configuration,
        file,
        sort_keys=False
    )


# ======================================================================
# 55. RESULTS JSON
# ======================================================================

results_json = {

    "individual_models":
        results_dataframe
        .to_dict(
            orient="records"
        ),

    "ensemble":
        ensemble_result,

    "ensemble_models":
        top_models
}


with open(

    PROJECT_DIR /
    "results.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(
        results_json,
        file,
        indent=4
    )


# ======================================================================
# 56. BEST MODEL
# ======================================================================

best_model_name = str(
    results_dataframe
    .iloc[0][
        "Model"
    ]
)


best_model_f1 = float(
    results_dataframe
    .iloc[0][
        "F1"
    ]
)


best_model_metrics = (
    results_dataframe
    .iloc[0]
    .to_dict()
)


with open(

    RESULTS_DIR /
    "best_model.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(
        best_model_metrics,
        file,
        indent=4
    )


print(
    "\nBest individual model:"
)

print(
    best_model_name
)


print(
    "Best individual F1:"
)

print(
    f"{best_model_f1:.4f}"
)


# ======================================================================
# 57. METADATA PKL
# ======================================================================

metadata = {

    "vocabulary":
        vocabulary,

    "maximum_length":
        MAX_LENGTH,

    "vocabulary_size":
        actual_vocab_size,

    "embedding_dimension":
        EMBEDDING_DIM,

    "best_model":
        best_model_name
}


with open(

    PROJECT_DIR /
    "sentiment_metadata.pkl",

    "wb"

) as file:

    pickle.dump(

        metadata,

        file,

        protocol=
            pickle.HIGHEST_PROTOCOL
    )


# ======================================================================
# 58. REQUIREMENTS.TXT
# ======================================================================

requirements_lines = [

    "tensorflow>=2.15",

    "numpy>=1.24",

    "pandas>=2.0",

    "matplotlib>=3.5",

    "scikit-learn>=1.3",

    "nltk>=3.8",

    "gensim>=4.3",

    "PyYAML>=6.0"
]


requirements_text = "\n".join(
    requirements_lines
)


with open(

    PROJECT_DIR /
    "requirements.txt",

    "w",

    encoding="utf-8"

) as file:

    file.write(
        requirements_text
    )


# ======================================================================
# 59. README
#
# NO TRIPLE-QUOTED STRING.
# This fixes the previous incomplete-input problem.
# ======================================================================

readme_lines = [

    "# Sentiment Analysis with Advanced NLP",

    "",

    "## Overview",

    "",

    "This project implements an end-to-end deep learning sentiment analysis system using the IMDB Movie Review dataset.",

    "",

    "The objective is to classify movie reviews as positive or negative and compare multiple neural network architectures.",

    "",

    "## Models Implemented",

    "",

    "1. Bidirectional LSTM",

    "2. GRU with Attention",

    "3. 1D Text CNN",

    "4. Hybrid CNN-LSTM",

    "5. Ensemble of the top-performing models",

    "",

    "## Dataset",

    "",

    "IMDB Movie Review Dataset",

    "",

    "- Negative = 0",

    "- Positive = 1",

    "",

    "## Dataset Split",

    "",

    "- Training = 70%",

    "- Validation = 15%",

    "- Testing = 15%",

    "",

    "## NLP Preprocessing",

    "",

    "- Lowercase conversion",

    "- HTML removal",

    "- URL removal",

    "- Special-character removal",

    "- Word-level tokenization",

    "- Character-level tokenization",

    "- Stopword analysis",

    "- Vocabulary creation",

    "- Sequence padding and truncation",

    "",

    "## Embeddings",

    "",

    "- Custom Word2Vec",

    "- Optional pretrained GloVe",

    "",

    "Large external GloVe files are excluded from the final ZIP.",

    "",

    "## Training",

    "",

    "- Adam optimizer",

    "- Binary cross-entropy loss",

    "- Class weighting",

    "- Dropout regularization",

    "- L2 regularization",

    "- Early stopping",

    "- Learning-rate reduction",

    "- Validation F1 model selection",

    "",

    "## Evaluation",

    "",

    "- Accuracy",

    "- Precision",

    "- Recall",

    "- F1 Score",

    "- ROC-AUC",

    "- Confusion Matrix",

    "- ROC Curve",

    "",

    "## Advanced Features",

    "",

    "- Top-three model ensemble",

    "- Error analysis",

    "- Confidence analysis",

    "- Word importance visualization",

    "- Real-time sentiment prediction",

    "",

    "## Output Directory",

    "",

    str(
        PROJECT_DIR
    ),

    "",

    "## Storage",

    "",

    "The final project is automatically compressed into a ZIP archive.",

    "",

    "If the ZIP becomes too large, non-best model weights are removed automatically to keep the submission below 50 MB."
]


readme_text = "\n".join(
    readme_lines
)


README_PATH = (
    PROJECT_DIR /
    "README.md"
)


with open(

    README_PATH,

    "w",

    encoding="utf-8"

) as file:

    file.write(
        readme_text
    )


print(
    "\nREADME created successfully."
)


# ======================================================================
# 60. PROJECT SUMMARY
# ======================================================================

summary_lines = [

    "SENTIMENT ANALYSIS WITH ADVANCED NLP",

    "=" * 60,

    "",

    "Dataset: IMDB Movie Reviews",

    "",

    "Total samples:",

    str(
        len(
            cleaned_texts
        )
    ),

    "",

    "Vocabulary size:",

    str(
        actual_vocab_size
    ),

    "",

    "Sequence length:",

    str(
        MAX_LENGTH
    ),

    "",

    "Embedding dimension:",

    str(
        EMBEDDING_DIM
    ),

    "",

    "Best individual model:",

    best_model_name,

    "",

    "Best individual F1:",

    f"{best_model_f1:.4f}",

    "",

    "Ensemble Accuracy:",

    f"{ensemble_accuracy:.4f}",

    "",

    "Ensemble Precision:",

    f"{ensemble_precision:.4f}",

    "",

    "Ensemble Recall:",

    f"{ensemble_recall:.4f}",

    "",

    "Ensemble F1:",

    f"{ensemble_f1:.4f}",

    "",

    "Ensemble ROC-AUC:",

    f"{ensemble_auc:.4f}"
]


summary_text = "\n".join(
    summary_lines
)


with open(

    PROJECT_DIR /
    "project_summary.txt",

    "w",

    encoding="utf-8"

) as file:

    file.write(
        summary_text
    )


# ======================================================================
# 61. REAL-TIME PREDICTION
# ======================================================================

best_model = (
    trained_models[
        best_model_name
    ]
)


def predict_sentiment(
    review
):

    cleaned_review = (
        clean_text(
            review
        )
    )


    sequence = vectorizer(
        [
            cleaned_review
        ]
    )


    probability = float(

        best_model.predict(
            sequence,
            verbose=0
        )[0][0]
    )


    if probability >= 0.5:

        sentiment = (
            "Positive"
        )

        confidence = (
            probability
        )

    else:

        sentiment = (
            "Negative"
        )

        confidence = (
            1
            -
            probability
        )


    return {

        "review":
            review,

        "sentiment":
            sentiment,

        "confidence":
            round(
                confidence,
                4
            ),

        "positive_probability":
            round(
                probability,
                4
            )
    }


# ======================================================================
# 62. DEMO PREDICTIONS
# ======================================================================

demo_reviews = [

    "This movie was absolutely amazing and I loved every minute of it.",

    "The film was boring and a complete waste of time.",

    "The acting was good but the story could have been better.",

    "One of the best movies I have seen.",

    "Terrible acting and a weak story."
]


demo_results = []


print(
    "\n" +
    "=" * 70
)


print(
    "REAL-TIME PREDICTIONS"
)


print(
    "=" * 70
)


for review in demo_reviews:

    result = (
        predict_sentiment(
            review
        )
    )


    demo_results.append(
        result
    )


    print(
        "\nReview:",
        review
    )


    print(
        "Prediction:",
        result[
            "sentiment"
        ]
    )


    print(
        "Confidence:",
        result[
            "confidence"
        ]
    )


pd.DataFrame(
    demo_results
).to_csv(

    RESULTS_DIR /
    "demo_predictions.csv",

    index=False
)


# ======================================================================
# 63. FOLDER SIZE FUNCTION
# ======================================================================

def folder_size_mb(
    folder
):

    total_size = 0


    for root, dirs, files in os.walk(
        folder
    ):

        for filename in files:

            filepath = (
                Path(root)
                /
                filename
            )


            if filepath.is_file():

                try:

                    total_size += (
                        filepath
                        .stat()
                        .st_size
                    )

                except OSError:

                    pass


    return (
        total_size
        /
        (1024 ** 2)
    )


# ======================================================================
# 64. FILE INVENTORY
# ======================================================================

file_inventory = []


for root, dirs, files in os.walk(
    PROJECT_DIR
):

    for filename in files:

        filepath = (
            Path(root)
            /
            filename
        )


        try:

            size_mb = (
                filepath
                .stat()
                .st_size
                /
                (1024 ** 2)
            )

        except OSError:

            size_mb = 0


        file_inventory.append(
            {
                "File":
                    str(
                        filepath.relative_to(
                            PROJECT_DIR
                        )
                    ),

                "Size_MB":
                    round(
                        size_mb,
                        4
                    )
            }
        )


inventory_dataframe = (
    pd.DataFrame(
        file_inventory
    )
)


if not (
    inventory_dataframe.empty
):

    inventory_dataframe = (
        inventory_dataframe
        .sort_values(
            "Size_MB",
            ascending=False
        )
        .reset_index(
            drop=True
        )
    )


inventory_dataframe.to_csv(

    PROJECT_DIR /
    "file_inventory.csv",

    index=False
)


print(
    "\nLargest generated files:"
)


if not (
    inventory_dataframe.empty
):

    print(

        inventory_dataframe
        .head(
            20
        )
        .to_string(
            index=False
        )
    )


# ======================================================================
# 65. ZIP FUNCTION
# ======================================================================

def create_zip():

    if ZIP_PATH.exists():

        ZIP_PATH.unlink()


    with zipfile.ZipFile(

        ZIP_PATH,

        "w",

        compression=
            zipfile.ZIP_DEFLATED,

        compresslevel=9

    ) as zip_file:


        for root, dirs, files in os.walk(
            PROJECT_DIR
        ):


            dirs[:] = [
                directory

                for directory
                in dirs

                if directory
                != "__pycache__"
            ]


            for filename in files:

                filepath = (
                    Path(root)
                    /
                    filename
                )


                # Do not include large GloVe
                if filename.startswith(
                    "glove."
                ):

                    continue


                if filename.endswith(
                    ".pyc"
                ):

                    continue


                if filename.endswith(
                    ".tmp"
                ):

                    continue


                archive_name = (

                    Path(
                        PROJECT_DIR.name
                    )
                    /
                    filepath.relative_to(
                        PROJECT_DIR
                    )
                )


                zip_file.write(

                    filepath,

                    arcname=
                        str(
                            archive_name
                        )
                )


    return (
        ZIP_PATH
        .stat()
        .st_size
        /
        (1024 ** 2)
    )


# ======================================================================
# 66. CREATE ZIP
# ======================================================================

print(
    "\nCreating project ZIP..."
)


zip_size = (
    create_zip()
)


print(
    f"Initial ZIP size: "
    f"{zip_size:.2f} MB"
)


# ======================================================================
# 67. AUTOMATIC SIZE REDUCTION
# ======================================================================

if zip_size >= 50:

    print(
        "\nZIP is above 50 MB."
    )

    print(
        "Removing non-best model weights..."
    )


    best_weight_filename = (
        f"{best_model_name}"
        ".weights.h5"
    )


    for weight_file in (
        MODELS_DIR.glob(
            "*.weights.h5"
        )
    ):

        if weight_file.name != (
            best_weight_filename
        ):

            try:

                weight_file.unlink()

                print(
                    "Removed:",
                    weight_file.name
                )

            except Exception as error:

                print(
                    "Could not remove:",
                    weight_file.name
                )

                print(
                    error
                )


    zip_size = (
        create_zip()
    )


    print(
        f"ZIP after model reduction: "
        f"{zip_size:.2f} MB"
    )


# ======================================================================
# 68. ADDITIONAL SIZE REDUCTION
# ======================================================================

if zip_size >= 50:

    print(
        "\nApplying additional "
        "size reduction..."
    )


    optional_files = [

        RESULTS_DIR /
        "predictions.csv.gz",

        RESULTS_DIR /
        "preprocessing_samples.csv"
    ]


    for optional_file in (
        optional_files
    ):

        if optional_file.exists():

            try:

                file_size = (
                    optional_file
                    .stat()
                    .st_size
                    /
                    (1024 ** 2)
                )


                if file_size > 1:

                    optional_file.unlink()

                    print(
                        "Removed:",
                        optional_file.name
                    )

            except Exception:

                pass


    zip_size = (
        create_zip()
    )


# ======================================================================
# 69. ZIP INFORMATION
# ======================================================================

zip_information = {

    "zip_path":
        str(
            ZIP_PATH
        ),

    "zip_size_mb":
        round(
            zip_size,
            4
        ),

    "below_50_mb":
        bool(
            zip_size < 50
        ),

    "best_model":
        best_model_name,

    "best_model_f1":
        round(
            best_model_f1,
            6
        ),

    "ensemble_f1":
        round(
            float(
                ensemble_f1
            ),
            6
        )
}


with open(

    RESULTS_DIR /
    "zip_information.json",

    "w",

    encoding="utf-8"

) as file:

    json.dump(
        zip_information,
        file,
        indent=4
    )


# Recreate ZIP so zip_information.json is included.
zip_size = (
    create_zip()
)


# ======================================================================
# 70. FINAL RESULTS
# ======================================================================

final_folder_size = (
    folder_size_mb(
        PROJECT_DIR
    )
)


plot_files = list(
    PLOTS_DIR.glob(
        "*.png"
    )
)


model_files = list(
    MODELS_DIR.glob(
        "*.weights.h5"
    )
)


print(
    "\n" +
    "=" * 70
)


print(
    "PROJECT COMPLETED SUCCESSFULLY"
)


print(
    "=" * 70
)


print(
    "\nProject directory:"
)


print(
    PROJECT_DIR
)


print(
    "\nProject folder size:"
)


print(
    f"{final_folder_size:.2f} MB"
)


print(
    "\nZIP file:"
)


print(
    ZIP_PATH
)


print(
    "\nZIP size:"
)


print(
    f"{zip_size:.2f} MB"
)


print(
    "\nBest individual model:"
)


print(
    best_model_name
)


print(
    "\nBest individual F1:"
)


print(
    f"{best_model_f1:.4f}"
)


print(
    "\nEnsemble Accuracy:"
)


print(
    f"{ensemble_accuracy:.4f}"
)


print(
    "\nEnsemble Precision:"
)


print(
    f"{ensemble_precision:.4f}"
)


print(
    "\nEnsemble Recall:"
)


print(
    f"{ensemble_recall:.4f}"
)


print(
    "\nEnsemble F1:"
)


print(
    f"{ensemble_f1:.4f}"
)


print(
    "\nEnsemble ROC-AUC:"
)


print(
    f"{ensemble_auc:.4f}"
)


print(
    "\nNumber of visualizations:"
)


print(
    len(
        plot_files
    )
)


print(
    "\nSaved model files:"
)


for model_file in model_files:

    size = (
        model_file
        .stat()
        .st_size
        /
        (1024 ** 2)
    )


    print(
        "-",
        model_file.name,
        f"{size:.2f} MB"
    )


print(
    "\n" +
    "=" * 70
)


if zip_size < 50:

    print(
        "SUCCESS: ZIP IS BELOW 50 MB"
    )

else:

    print(
        "WARNING: ZIP IS ABOVE 50 MB"
    )


print(
    "=" * 70
)


print(
    "\nAll project results are saved in:"
)


print(
    r"C:\Users\sagni\Downloads\Sentiment Analysis with Advanced NLP"
)


print(
    "\nSubmission ZIP:"
)


print(
    r"C:\Users\sagni\Downloads\Sentiment_Analysis_with_Advanced_NLP.zip"
)


print(
    "\nDone."
)
```
